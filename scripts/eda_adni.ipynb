```python
def eda(df):
    """
    Resumen de cada columna de un DataFrame.

    Para todas las columnas calcula nulos, valores únicos y el valor más
    repetido. Para las numéricas añade media, desviación, mínimo, mediana
    y máximo.

    Returns
    -------
    numericas : pandas.DataFrame
        Una fila por columna numérica.
    no_numericas : pandas.DataFrame
        Una fila por columna no numérica (texto, categorías, booleanos).
    """
    rows = []
    for col in df.columns:
        s = df[col]
        # Los booleanos los tratamos como no numéricos (True/False son categorías)
        is_num = pd.api.types.is_numeric_dtype(s) and not pd.api.types.is_bool_dtype(s)
        moda = s.mode(dropna=True)
        frec = (s == moda.iloc[0]).sum() if len(moda) > 0 else 0

        row = {
            "column name": col,
            "type": "numerical" if is_num else "non numerical",
            "n_nulls": s.isna().sum(),
            "%_nulls": round(s.isna().mean() * 100, 1),
            "n_unique": s.nunique(dropna=True),
            "most_frequent": moda.iloc[0] if len(moda) > 0 else None,
            "frequency_most_frequent": frec,
            "%_most_frequent": round(frec / len(s) * 100, 1),
        }
        if is_num:
            row.update({
                "media": s.mean(),
                "std": s.std(),
                "min": s.min(),
                "mediana": s.median(),
                "max": s.max(),
            })
        else:
            values = sorted(s.dropna().astype(str).unique())
            row["values"] = values if len(values) <= 20 else values[:20] + ["..."]
        rows.append(row)

    summary = pd.DataFrame(rows).set_index("column name")
    numerical = summary[summary["type"] == "numerical"].drop(columns=["type", "values"], errors="ignore")
    non_numerical = summary[summary["type"] == "non numerical"].drop(
        columns=["type", "media", "std", "min", "mediana", "max"], errors="ignore"
    )

    return numerical, non_numerical
```

## Create final cohort

```python
df_adni_v1 = df_adni.copy()

df_adni_v1["DX_bl_binary"] = df_adni_v1["DX_bl"].replace({"LMCI": "AD", "EMCI": "AD", "SMC": "CN"})
df_adni_v1["DX_bl_binary"] = df_adni_v1["DX_bl_binary"].fillna("Unknown")

df_adni_v1["DX_binary"] = df_adni_v1["DX"].replace({"Dementia": "AD", "MCI": "AD"})
df_adni_v1["DX_binary"] = df_adni_v1["DX_binary"].fillna("Unknown")

ax = df_adni_v1["DX_binary"].value_counts(dropna=False).plot(kind="bar")
ax.bar_label(ax.containers[0])
```

```python
df_adni_v2 = df_adni_v1.copy()

df_adni_v2["DX_binary"] = np.where(
    (df_adni_v2["DX_binary"] == 'Unknown') & (df_adni_v2["DX_bl_binary"] == "AD"),
    "AD",
    df_adni_v2["DX_binary"],
) # fill DX_binary with AD if DX_bl_binary is AD and DX_binary is Unknown, AD subjects can't go back to CN


ax = df_adni_v2["DX_binary"].value_counts(dropna=False).plot(kind="bar")
ax.bar_label(ax.containers[0])
```

```python
df_adni_v3 = df_adni_v2.copy()

df_adni_v3['DX_binary'] = np.where(
    (df_adni_v3['VISCODE'] == 'bl') & (df_adni_v3['DX_binary'] == 'Unknown') & (df_adni_v3['DX_bl_binary'] != 'Unknown'),
    df_adni_v3['DX_bl_binary'],
    df_adni_v3['DX_binary']
) # If the visit is the baseline and the DX_bl is informed, fill DX with it

ax = df_adni_v3["DX_binary"].value_counts(dropna=False).plot(kind="bar")
ax.bar_label(ax.containers[0])
```

# Explore metadata from nifti files

## Create Dataframe from json's files

```python
import json
import re
from pathlib import Path
import pandas as pd

root = Path("/pool/home/AD_Multimodal/ADNI/ADNI-BIDS")

rows = []
for p in root.rglob("*_T1w.json"):
    if p.name.startswith("._"):
        continue
    try:
        with open(p, "r", encoding="utf-8") as f:
            d = json.load(f)
    except (UnicodeDecodeError, json.JSONDecodeError) as e:
        continue

    # Treu sub i ses del nom del fitxer BIDS (ex: sub-002S0295_ses-M00_T1w.json)
    sub = re.search(r"sub-([^_/]+)", p.name)
    ses = re.search(r"ses-([^_/]+)", p.name)
    run = re.search(r"run-(\d+)", p.name)
    d["subject"] = sub.group(1) if sub else None
    d["session"] = ses.group(1) if ses else None
    d["run"] = int(run.group(1)) if run else 1   # sense run-XX = una sola adquisició
    d["path"] = str(p)

    rows.append(d)

df_metadata = pd.DataFrame(rows)

df_metadata = (
    df_metadata
    .sort_values(["subject", "session", "run"])
    .drop_duplicates(subset=["subject", "session"], keep="last") # keep only last run
    # .reset_index(drop=True)
)

print(df_metadata.shape)
```

```python
df_metadata.head()
```

```python
# convert to text list and dics

list_cols = [c for c in df_metadata.columns
             if df_metadata[c].apply(lambda x: isinstance(x, (list, dict))).any()]
print(list_cols)

for c in list_cols:
    df_metadata[c] = df_metadata[c].apply(
        lambda x: json.dumps(x) if isinstance(x, (list, dict)) else x
    )

numeric_metadata, non_numeric_metadata = eda(df_metadata)

display(numeric_metadata)
display(non_numeric_metadata)
```

```python
import nibabel as nib

def plot_grid(mask, n=20):
    sel = df_metadata.loc[mask, ["subject", "session", "run", "BodyPartExamined", "path"]].head(n)

    fig, axes = plt.subplots(4, 5, figsize=(18, 15))
    for ax, (_, r) in zip(axes.flat, sel.iterrows()):
        nii = Path(r["path"].replace(".json", ".nii.gz"))
        if not nii.exists():
            nii = Path(r["path"].replace(".json", ".nii"))
        try:
            img = nib.as_closest_canonical(nib.load(nii))
            data = img.get_fdata()
            sl = data[:, :, data.shape[2] // 2]   # corte axial central
            vmax = np.percentile(sl, 99)
            ax.imshow(sl.T, cmap="gray", origin="lower", vmin=0, vmax=vmax)
        except Exception as e:
            ax.text(0.5, 0.5, f"Error:\n{e}", ha="center", va="center", fontsize=8, wrap=True)
        ax.set_title(f"{r['subject']} {r['session']} run{r['run']}\n{r['BodyPartExamined']}", fontsize=9)
        ax.axis("off")

    for ax in axes.flat[len(sel):]:
        ax.axis("off")
    plt.tight_layout()
    plt.show()
```

```python
mask_brain = df_metadata["BodyPartExamined"].isin(["BRAIN", "HEAD"])
plot_grid(mask_brain)
```

```python
mask_other = ~df_metadata["BodyPartExamined"].isin(["BRAIN", "HEAD"]) & df_metadata["BodyPartExamined"].notna()
plot_grid(mask_other)
```
